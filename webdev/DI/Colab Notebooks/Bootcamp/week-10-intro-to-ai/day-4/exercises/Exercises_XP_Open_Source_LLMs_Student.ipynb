## Exercise 1: Open Source Levels Reflection
**Use:** Fully Open, Weights Released, Architecture Only; model components; fine-tuning/domain adaptation; healthcare compliance.

**Step-by-step**
1) Gather definitions for the three openness levels.  
2) Identify key characteristics (what is open; what you can/cannot do).  
3) Compare side by side in a table (what's open vs. impact).  
4) Write a 3?5 sentence comparative paragraph.  
5) Answer the healthcare prompt (1?2 sentences) about retraining on clinical data.

**Deliverables:** paragraph + healthcare answer.

```
# capture definitions and key characteristics
open_source_levels = {
    "Fully Open": {
        "definition": "everything is publicly released - weights, training code, training data, and architecture",
        "what_is_open": ["model weights and architecture", "training data and training code"],
        "what_you_can_do": ["fine-tune on your own data", "fully reproduce and modify the model"],
        "what_you_cannot_do": ["nothing is really restricted with this level of openness"],
    },
    "Weights Released": {
        "definition": "the trained model weights and architecture are shared but training data and full training code are not",
        "what_is_open": ["model weights", "model architecture/config"],
        "what_you_can_do": ["run inference locally", "fine-tune on your own dataset"],
        "what_you_cannot_do": ["reproduce the original training run", "access or audit the training data"],
    },
    "Architecture Only": {
        "definition": "only the design and architecture of the model is public, no weights or training data",
        "what_is_open": ["model design and layer structure", "hyperparameter choices from the paper"],
        "what_you_can_do": ["build your own version of the architecture", "train from scratch if you have data and compute"],
        "what_you_cannot_do": ["use the model directly without training it yourself", "fine-tune since there are no weights to start from"],
    },
}
open_source_levels
```

```
# fill the comparison table (keep markdown formatting)
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | weights, code, training data, architecture | can fully reproduce, modify, and retrain from scratch |
| Weights Released | weights and architecture only | can fine-tune and run inference but can't reproduce full training |
| Architecture Only | just the design/structure | have to train from scratch, no starting point for fine-tuning |"""
print(comparison_table)
```

```
# 3-5 sentence comparative paragraph
comparative_paragraph = """Fully open models give you the most flexibility since everything from the weights to the training data is available, meaning you can retrain, audit, and modify pretty much anything. Weights released models like Mistral or Llama 2 are the most common in practice - you get the model ready to use and fine-tune, but you can't see what data it was trained on. Architecture only models are mostly just research contributions since you'd need your own data and a lot of compute to actually build something useful from them. For most real projects, weights released is the sweet spot because you get a working model without needing to start from zero."""
print(comparative_paragraph)
```

```
# 1-2 sentence healthcare prompt answer
healthcare_prompt_answer = """For healthcare, you'd want a fully open model so you can retrain it on clinical data and audit what it was originally trained on - using a weights-released model means you can't verify if the pretraining data had biases or errors that could cause problems in a medical setting. Compliance requirements like HIPAA also make full transparency over training data and model behavior way more important than in other domains."""
print(healthcare_prompt_answer)
```
