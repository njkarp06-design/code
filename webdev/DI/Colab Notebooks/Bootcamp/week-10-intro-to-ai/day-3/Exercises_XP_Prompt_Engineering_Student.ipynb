## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# TODO: write the full multi-part prompt
quiz_prompt = (
    "You are a middle school teacher creating content for 7th grade students. "
    "Using only the article about volcanic eruptions below, do the following:\n\n"
    "1. Write a 2 bullet point summary of the main facts using simple words that 11-13 year olds can understand\n"
    "2. Create 3 multiple choice questions based on the article. Each question needs 1 correct answer and 2 wrong but believable options\n\n"
    "Use a friendly and encouraging tone. Format the output so it can be copied straight into Google Slides "
    "- use clear headers for the Summary section and the Quiz section.\n\n"
    "[paste article here]"
)
quiz_prompt
```

```python
# TODO: explain three improvements over the vague prompt
improvements = [
    'specifies the exact format needed (2 bullets + 3 MCQs with 1 correct and 2 wrong) so theres no ambiguity',
    'sets the reading level and tone for the right age group instead of just saying kids',
    'includes output format instructions so it can go straight into Google Slides without extra editing',
]
improvements
```
