# Daily Challenge: Precision Prompting for Control & Output Quality
Working through the full prompt development cycle — design, evaluate, fix hallucination risk, paraphrase, and quote.

## Step 1: Craft a Prompt That Controls Tone, Format, and Length
Rewrite the remote access policy text in a friendly tone, bullet points, paraphrased, under 75 words.

```python
original_text = (
    "Employees must ensure that all remote access to internal systems is established via the approved "
    "secure VPN. Under no circumstances should unsecured connections or personal devices lacking "
    "endpoint protection be used to access proprietary data or sensitive communications."
)

step1_prompt = (
    "You are writing content for an internal employee newsletter. Rewrite the text below in a "
    "friendly, clear tone that any employee can understand - not just technical staff. "
    "Use bullet points to organize the information. Paraphrase the original, do not quote it directly. "
    "Keep the total response under 75 words.\n\n"
    f"Text: {original_text}"
)
print(step1_prompt)
```

## Step 2: Evaluate the Output
Review the generated response against the five quality criteria.

```python
output_evaluation = {
    'relevance': 'yes, core message about VPN and avoiding unsecured devices came through',
    'clarity': 'much easier to read than the original, non-technical staff would follow it fine',
    'structure': 'bullet points present, formatted as requested',
    'tone': 'friendly and approachable, fits internal employee comms',
    'length_under_75_words': True,
    'factual_accuracy': 'stayed close to the source, no extra details were added',
}
output_evaluation
```

## Step 3: Detect and Mitigate Hallucinations
If the model added content not in the original (extra tools, policy terms), add a constraint to lock it down.

```python
# added a line at the end to stop it going beyond what was actually in the text
step3_prompt_revised = (
    "You are writing content for an internal employee newsletter. Rewrite the text below in a "
    "friendly, clear tone that any employee can understand - not just technical staff. "
    "Use bullet points to organize the information. Paraphrase the original, do not quote it directly. "
    "Keep the total response under 75 words. "
    "Only paraphrase the content in the input text. Do not add new recommendations, technologies, "
    "or policies that are not mentioned in the original.\n\n"
    f"Text: {original_text}"
)
print(step3_prompt_revised)
```

## Step 4: Paraphrasing Deep Dive
Rewrite the same policy for a junior intern audience — plain language, max 4 bullets, supportive tone, no jargon.

```python
step4_intern_prompt = (
    "You are writing a quick heads-up for new interns just starting at the company. "
    "Using only the text below, rewrite the key points in plain everyday language - "
    "no corporate or legal words. Use no more than 4 short bullet points. "
    "Keep the tone friendly and supportive, like a colleague giving a new person a quick tip.\n\n"
    f"Text: {original_text}"
)
print(step4_intern_prompt)
```

## Step 5: Quote Extraction Variant
Extract the best direct quote from the original, then answer the follow-up questions.
